# START HERE – Single-Cell Best Practices Bootcamp

Welcome! This notebook is your control panel. Click the day you want, open the linked notebook, then press `Run` from top to bottom just like the DeepTCR new implementation experience.


## Quick instructions
1. **Open Jupyter or VS Code.** Navigate to the `Day_SC_Best_Practices` folder.
2. **Click one of the day links below.** Each link opens another notebook (Day 01 → Day 10).
3. **Run cells in order.** Every cell already contains the code—you only press the ▶️ button.
4. **Follow the markdown cues.** Before each cell you’ll see a short “what / why / how” explanation in plain English.
5. **Only install once.** The very first code cell in every notebook is a commented `%pip install …` line—uncomment it if you’re missing a package, then re-run.
