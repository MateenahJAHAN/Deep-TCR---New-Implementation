## Step 2: Reuse Your Day 3 + Day 5 Helper Functions

Before looping through lots of files, make sure the two building blocks below are clear:

1. **`clean_patient_data(df)`** – copy/paste (or rewrite) your Day 3 logic that kept productive, valid-length sequences.
2. **`encode_sequence(seq)`** – reuse your Day 5 encoder so every amino-acid string becomes a `(40, 20)` array.

> _If you can explain what goes in and what comes out of each helper, the multi-patient loop that follows will feel like plugging Lego bricks together._

**In plain words:** `clean_patient_data` = “wash the spreadsheet so only good rows remain,” `encode_sequence` = “turn each cleaned row into numbers the model can read.”

**Paper link:** Methods – *Data curation* (`docs/paper.pdf`, Sidhom et al., 2022) lists the exact same filters: productive sequences only, valid amino-acid letters, and length ≤ 40. You are literally recreating the authors' preprocessing pipeline here.

````python
# Define cleaning function (from Day 3)
# This function cleans one patient's data
# 
# **Remember from Day 3:**
# - We learned how to filter to productive sequences
# - We learned how to remove invalid amino acids
# - We learned how to filter by length (10-25)
# - Today we're putting it all together in a function!

def clean_patient_data(df):
    """
    Clean patient TCR data
    
    Steps:
    1. Filter to productive sequences
    2. Remove invalid amino acids
    3. Filter by length (10-25)
    4. Return cleaned DataFrame
    """
    # Filter to productive
    productive = df[df['sequenceStatus'] == 'In'].copy()
    
    # Remove invalid amino acids
    valid_aa = set('ACDEFGHIKLMNPQRSTVWY')
    is_valid = productive['aminoAcid'].apply(lambda x: all(c in valid_aa for c in str(x)))
    valid = productive[is_valid].copy()
    
    # Filter by length
    lengths = valid['aminoAcid'].str.len()
    filtered = valid[(lengths >= 10) & (lengths <= 25)].copy()
    
    return filtered

print("✓ Defined cleaning function")
print("  This function cleans data like we did in Day 3!")
````

````python
# Define encoding function (from Day 5)
# This function encodes sequences
# 
# **Remember from Day 5:**
# - We learned how to create one-hot encoding
# - We learned how to encode one amino acid → vector of 20
# - We learned how to encode a full sequence → matrix (length, 20)
# - Today we're using that function to encode ALL sequences!

# Amino acid alphabet (from Day 5)
AA_ALPHABET = 'ACDEFGHIKLMNPQRSTVWY'
aa_to_idx = {aa: idx for idx, aa in enumerate(AA_ALPHABET)}

def encode_sequence(sequence, max_length=40, aa_to_idx=aa_to_idx):
    """
    Encode amino acid sequence to one-hot matrix
    """
    encoded = np.zeros((max_length, 20))
    for i, aa in enumerate(sequence):
        if i < max_length:  # Don't exceed max_length
            encoded[i, aa_to_idx[aa]] = 1
    return encoded

print("✓ Defined encoding function")
print("  This function encodes sequences like we did in Day 5!")
````

````python
# Process each patient
# We'll store cleaned DataFrames and encoded arrays
# 
# Steps for each file:
#   1. Load file: pd.read_csv(file, sep='\t')
#   2. Clean data: clean_patient_data(df)
#   3. Get sequences: cleaned['aminoAcid'].tolist()
#   4. Encode sequences: loop and encode each
#   5. Store results

# Initialize lists to store results
patient_data = []  # Store cleaned DataFrames
patient_encoded = []  # Store encoded arrays
patient_names = []  # Store patient names

# ✅ Starter solution
# If you haven't written the TODO yet, we'll process files for you.
if selected_files is not None and len(selected_files) > 0:
    for file_path in selected_files:
        try:
            print(f"\nProcessing: {file_path.name}")
            
            # ✅ Starter: Load the file
            df = pd.read_csv(file_path, sep='\t')
            
            # ✅ Starter: Clean the data
            cleaned = clean_patient_data(df)
            
            if cleaned is not None and len(cleaned) > 0:
                print(f"  Cleaned: {len(cleaned)} sequences")
                
                # ✅ Starter: Get sequences
                sequences = cleaned['aminoAcid'].tolist()
                
                # ✅ Starter: Encode sequences (limit to first 1000 for speed)
                encoded_list = []
                for seq in sequences[:1000]:  # Limit for speed
                    encoded_seq = encode_sequence(seq)
                    encoded_list.append(encoded_seq)
                
                # Convert to numpy array
                if len(encoded_list) > 0:
                    encoded_array = np.array(encoded_list)
                    
                    # Store results
                    patient_data.append(cleaned)
                    patient_encoded.append(encoded_array)
                    patient_names.append(file_path.stem)
                    
                    print(f"  Encoded: {encoded_array.shape} (sequences, positions, amino_acids)")
                else:
                    print("  ⚠️ No sequences to encode")
            else:
                print("  ⚠️ No sequences after cleaning")
        except Exception as e:
            print(f"  ⚠️ Error processing {file_path.name}: {e}")
    
    print(f"\n✓ Processed {len(patient_data)} patients")
    print("\n💡 Tip: Replace the starter code above with your own implementation once you're ready.")
else:
    print("⚠️ Complete the TODO in previous cell to select files first!")
    patient_data = []
    patient_encoded = []
    patient_names = []

# TODO: Replace starter code above with your own implementation
# Steps:
#   1. Loop through selected_files
#   2. Load each file with pd.read_csv()
#   3. Clean with clean_patient_data()
#   4. Get sequences from cleaned['aminoAcid']
#   5. Encode each sequence with encode_sequence()
#   6. Store in patient_data, patient_encoded, patient_names
````

````python
# Load response labels and align with patient names
label_path = Path("../data/DeepTCR_Cancer-master/Data/yost/response.csv")
if not label_path.exists():
    raise FileNotFoundError(f"Could not find labels CSV at {label_path}. Double-check your data path.")

responses = pd.read_csv(label_path)
responses['patient_id'] = responses['Patient'].str.strip().str.lower()
responses['label'] = responses['Response'].str.strip().str.lower().map({'yes': 1, 'no': 0})
label_map = dict(zip(responses['patient_id'], responses['label']))

patient_labels = []
missing_patients = []
for name in patient_names:
    patient_id = str(name).split('_')[0].lower()
    label = label_map.get(patient_id)
    if label is None:
        missing_patients.append(patient_id)
        patient_labels.append(None)
    else:
        patient_labels.append(label)

if missing_patients:
    print("⚠️ Missing labels for:", sorted(set(missing_patients)))
    print("   Verify response.csv contains those patient IDs before proceeding.")
else:
    print(f"✓ Matched labels for all {len(patient_labels)} patients")
    print(f"  Responders: {sum(patient_labels)} | Non-responders: {len(patient_labels)-sum(patient_labels)}")
````
