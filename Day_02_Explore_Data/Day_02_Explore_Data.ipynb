## 🧬 Biology Context: Understanding TCR-seq Data

### What Is a TCR Sequence?

**Simple explanation:**
- A TCR sequence is a string of amino acids (like "CASSLAPGATNEKLFF")
- Each sequence is unique to one T cell
- **Think:** Like a barcode that identifies each T cell

**Why amino acids?**
- T cells recognize proteins (which are made of amino acids)
- The TCR sequence determines what protein the T cell recognizes
- **Biology:** 20 standard amino acids (A, C, D, E, F, G, H, I, K, L, M, N, P, Q, R, S, T, V, W, Y)

### What Are V, D, J Genes?

**Simple explanation:**
- TCR genes are made by combining V, D, and J segments
- Like mixing and matching puzzle pieces
- **V gene:** Variable segment (many options)
- **D gene:** Diversity segment (fewer options)
- **J gene:** Joining segment (fewer options)
- **Biology:** This recombination creates diversity - millions of possible TCRs!

### What Does "Productive" Mean?

**Simple explanation:**
- "In" frame = productive (the sequence works correctly)
- "Out" frame = non-productive (the sequence doesn't work)
- **Biology:** Only productive sequences can actually recognize antigens
- **Why it matters:** We only care about sequences that actually function!

### What Are Read Counts?

**Simple explanation:**
- How many times this sequence was seen in the sequencing
- Higher count = sequence is more common in this patient
- **Biology:** Common sequences = more T cells with that barcode
- **Why it matters:** Frequent sequences might be important for immune response

## 💻 Computer Science Context: Working with Data

### What Is a TSV File?

**Simple explanation:**
- TSV = Tab-Separated Values
- Like CSV but uses tabs instead of commas
- **CS reason:** Some data contains commas, so tabs are safer
- **How to read:** `pd.read_csv(file, sep='\t')` - the `sep='\t'` tells pandas "use tabs"

### What Is a DataFrame?

**Simple explanation:**
- A DataFrame is like an Excel spreadsheet
- Rows = individual TCR sequences
- Columns = different pieces of information (sequence, V gene, count, etc.)
- **CS reason:** Easy to work with structured data
- **What you know:** Just like `pd.read_csv()` creates a DataFrame!

### Why Filter Data?

**CS reason:**
- Remove bad/irrelevant data
- Focus on what matters
- Make analysis easier
- **In this project:** We filter to productive sequences only
