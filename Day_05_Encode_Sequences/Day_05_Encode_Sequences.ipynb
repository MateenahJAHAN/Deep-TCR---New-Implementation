## 🤔 Questions to Think About

1. **Why do we need to convert strings to numbers?**
   - Computers work with numbers, not strings
   - Machine learning models need numeric input
   - Think: Like `pd.get_dummies()` converts categories to numbers

2. **What is one-hot encoding?**
   - One 1, rest are 0s
   - Like `pd.get_dummies()` but for sequences
   - Each position gets a vector of 20 numbers (one for each amino acid)

3. **What shape will the encoded array be?**
   - One sequence: (length, 20) - length positions, 20 amino acids
   - All sequences: (num_sequences, max_length, 20)
   - After padding: (num_sequences, 40, 20) - padded to 40 positions
