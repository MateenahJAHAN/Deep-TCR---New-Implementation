## 🧬 Biology Context: Why Convert Sequences to Numbers?

### What Are Amino Acids?

**Simple explanation:**
- Amino acids are building blocks of proteins
- There are 20 standard amino acids
- Each amino acid has a letter: A, C, D, E, F, G, H, I, K, L, M, N, P, Q, R, S, T, V, W, Y
- **Think:** Like letters in an alphabet, but for proteins!

### Why Convert to Numbers?

**Biology reason:**
- TCR sequences are strings of amino acids (like "CASSLAPG")
- But computers work with numbers, not strings
- **Why it matters:** Machine learning models need numeric input!

### What Is One-Hot Encoding?

**Simple explanation:**
- Each amino acid becomes a vector of 20 numbers
- One position is 1, rest are 0
- **Example:** 'A' → [1, 0, 0, 0, ..., 0] (1 at position 0)
- **Think:** Like a light switch - only one is "on"!

**Why one-hot?**
- Preserves information about which amino acid it is
- Easy for computers to process
- **Biology:** Each position tells us "this amino acid is here"

## 💻 Computer Science Context: Encoding Data

### Why Encode?

**CS reason:**
- Machine learning models need numeric input
- Strings can't be used directly in math operations
- **What you know:** Like `pd.get_dummies()` converts categories to numbers!

### What Is One-Hot Encoding?

**CS explanation:**
- **One-hot** = one 1, rest are 0s
- Like `pd.get_dummies()` but for sequences
- **Example:** 'A' → [1, 0, 0, ...], 'C' → [0, 1, 0, ...]
- **Why:** Preserves categorical information as numbers

### Understanding Array Shapes

**CS concept:**
- **Shape** = dimensions of an array
- `(100, 40, 20)` = 100 sequences, 40 positions, 20 amino acids
- **Why it matters:** Shapes must match for operations!

**Shape transformations:**
1. Original: `(sequences, columns)` - DataFrame
2. Encoded: `(sequences, max_length, 20)` - 3D array
3. Flattened: `(sequences, max_length * 20)` - 2D array
4. Final: `(sequences, features)` - Ready for ML!

## Step 1: Understand One-Hot Encoding

**Biology context:**
- We have 20 amino acids: A, C, D, E, F, G, H, I, K, L, M, N, P, Q, R, S, T, V, W, Y
- Each amino acid needs to become a number
- **One-hot encoding:** One position is 1, rest are 0

**CS context:**
- Create a mapping: amino acid → index
- Create a vector of 20 zeros
- Set the corresponding position to 1
- **What you know:** Like creating a lookup table!

**Your Task:** Create a mapping and encode one amino acid

````python
import numpy as np

AA_ALPHABET = "ACDEFGHIKLMNPQRSTVWY"
print(f"✓ Amino acids: {AA_ALPHABET}")
print(f"  Total: {len(AA_ALPHABET)} amino acids")
print("\nBiology: These are the building blocks of proteins!")
print("CS: We'll convert each letter to a number.")

if ("aa_to_idx" not in globals()) or (aa_to_idx is None):
    aa_to_idx = {aa: idx for idx, aa in enumerate(AA_ALPHABET)}
    print("\n✓ Starter mapping created for you (feel free to recreate it yourself).")
else:
    print("\n✓ Using the mapping you created above.")

print("\nAmino acid to index mapping (first 5 entries):")
for aa in AA_ALPHABET[:5]:
    print(f"  {aa} -> {aa_to_idx[aa]}")
````

````python
encoded_A = np.zeros(len(AA_ALPHABET), dtype=int)
encoded_A[aa_to_idx['A']] = 1

print("\n✓ Encoded 'A':")
print(f"  Vector: {encoded_A}")
print(f"  Shape: {encoded_A.shape}")
print(f"  Sum: {encoded_A.sum()} (should be 1 - one 'on', rest 'off')")
print("\nBiology: This tells us 'amino acid A is at this position'")
print("CS: One-hot encoding - one 1, rest 0s!")
````

````python
encoded_C = np.zeros(len(AA_ALPHABET), dtype=int)
encoded_C[aa_to_idx['C']] = 1

print("\n✓ Encoded 'C':")
print(f"  Vector: {encoded_C}")
print(f"  Shape: {encoded_C.shape}")
print(f"  Sum: {encoded_C.sum()}")

print("\nComparison vs 'A':")
print(f"  Dot product (should be 0): {encoded_A @ encoded_C}")
print("  Different amino acids have orthogonal one-hot vectors.")
````

## Step 2: Encode One Sequence

**Biology context:**
- A TCR sequence is a string of amino acids (like "CASSLAPG")
- Each position has one amino acid
- We need to encode the entire sequence

**CS context:**
- Loop through each position in the sequence
- Encode each amino acid
- Stack them into a matrix
- **Result:** `(sequence_length, 20)` array

**Your Task:** Write a function to encode a full sequence

````python
def encode_sequence(sequence, max_length=40, aa_map=None):
    """Encode a TCR amino acid sequence into a (max_length, 20) one-hot matrix."""
    if aa_map is None:
        aa_map = aa_to_idx
    encoded = np.zeros((max_length, len(AA_ALPHABET)), dtype=int)
    for i, aa in enumerate(sequence):
        if i >= max_length:
            break
        if aa not in aa_map:
            raise ValueError(f"Unknown amino acid '{aa}' encountered")
        encoded[i, aa_map[aa]] = 1
    return encoded

# Test your function
test_seq = "CASSLAPG"
print(f"Testing with sequence: {test_seq}")
print(f"  Length: {len(test_seq)} amino acids")
encoded = encode_sequence(test_seq, max_length=40)

print("\n✓ Encoded sequence:")
print(f"  Shape: {encoded.shape}")
print("  Expected: (40, 20) - 40 positions, 20 amino acids per position")
print(f"\n  First position (C): {encoded[0]}")
print(f"  Second position (A): {encoded[1]}")
print("  Biology: Each row tells us which amino acid is at that position.")
print("  CS: This matrix is ready to feed into a neural network.")
````

## Step 3: Encode V/D/J Genes

**Biology context:**
- V, D, J genes determine part of the TCR structure
- Each TCR uses one V gene, one D gene (sometimes), one J gene
- **Why it matters:** Gene usage patterns predict response!

**CS context:**
- Similar to amino acid encoding, but simpler
- One V gene → one vector
- **What you know:** Just like encoding categories with `pd.get_dummies()`!

**Your Task:** Encode V genes (D and J are similar)

````python
import pandas as pd
from pathlib import Path

# Locate data and load a sample patient file
_aa_data_dir = Path("../data/DeepTCR_Cancer-master/Data/yost/data")
if not _aa_data_dir.exists():
    for alt in [Path("data/DeepTCR_Cancer-master/Data/yost/data"), Path("./data/DeepTCR_Cancer-master/Data/yost/data")]:
        if alt.exists():
            _aa_data_dir = alt
            print(f"✓ Found data at: {alt}")
            break

sample_file = sorted(_aa_data_dir.glob("*.tsv"))[0]
df = pd.read_csv(sample_file, sep='\t')
productive = df[df['sequenceStatus'] == 'In'].copy()
unique_v_genes = sorted(productive['vGeneName'].dropna().unique())

print(f"✓ Loaded {sample_file.name}: {len(productive):,} productive sequences")
print(f"  Found {len(unique_v_genes)} unique V genes")
print(f"  Examples: {unique_v_genes[:5]}")

v_gene_to_idx = {gene: idx for idx, gene in enumerate(unique_v_genes)}
print("\n✓ Created V-gene mapping with", len(v_gene_to_idx), "entries")
data_dir = _aa_data_dir
````

````python
def encode_v_gene(v_gene, mapping):
    """One-hot encode a V gene using the mapping from above."""
    if v_gene not in mapping:
        raise ValueError(f"Unknown V gene: {v_gene}")
    vec = np.zeros(len(mapping), dtype=int)
    vec[mapping[v_gene]] = 1
    return vec

# Test the helper
sample_v = productive['vGeneName'].dropna().iloc[0]
encoded_v = encode_v_gene(sample_v, v_gene_to_idx)
print(f"Testing V gene encoding for: {sample_v}")
print(f"  Vector length: {len(encoded_v)} (should match number of V genes)")
print(f"  Ones in vector: {encoded_v.sum()} (should be 1)")
````

## Step 4: Encode All Sequences for One Patient

- We need to encode all of them
- **Result:** Array of shape `(num_sequences, 40, 20)`

**CS context:**

````python
# Encode amino-acid sequences into one-hot tensors
sequences = productive['aminoAcid'].dropna().tolist()
print(f"✓ Collected {len(sequences):,} sequences to encode")

max_examples = min(200, len(sequences))  # keep runtime manageable
encoded_sequences = np.stack([encode_sequence(seq, max_length=40) for seq in sequences[:max_examples]])

print("\n✓ Encoded", encoded_sequences.shape[0], "sequences")
print(f"  Tensor shape: {encoded_sequences.shape} -> (batch, positions, amino_acids)")
````
