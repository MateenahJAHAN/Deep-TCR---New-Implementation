## 🧬 Biology Context: What Are We Studying?

### What Are T Cells?

**Simple explanation:**
- T cells are immune cells (like soldiers in your body)
- They patrol your body looking for problems (like cancer cells)
- When they find a problem, they attack it
- **Think:** Like security guards that recognize bad guys

### What Is a TCR (T-Cell Receptor)?

**Simple explanation:**
- Each T cell has a unique "barcode" called a TCR
- This barcode determines what the T cell recognizes
- Like: "This T cell recognizes cancer cell type X"
- **Think:** Like a key that only fits certain locks

### What Is TCR-seq?

**Simple explanation:**
- TCR-seq = T-cell receptor sequencing
- Scientists read all the barcodes from a patient's T cells
- One patient might have 50,000 different T cell barcodes!
- **Think:** Like scanning all the barcodes in a store

### Why Does This Matter?

**The research question:**
- Can we predict if a patient will respond to immunotherapy?
- By looking at their T cell barcodes (TCR sequences)
- **Paper finding:** Yes! Certain patterns predict response

## Step 1: Check Your Python Version

First, let's make sure you have Python 3.7 or higher.

**Why Python 3.7+?**
- Newer packages require Python 3.7 or higher
- Some features we'll use weren't available in older versions
- **CS reason:** Language features and package compatibility

**Your Task:** Run the cell below to check your Python version.

```python
# Check your Python version
# This code uses the 'sys' module (built into Python)
# sys.version_info gives us the version number

import sys

# Print full version string
print(f"Python version: {sys.version}")

# Print just the version number (major.minor)
print(f"Version number: {sys.version_info.major}.{sys.version_info.minor}")

# Check if version is 3.7 or higher
# sys.version_info is a tuple like (3, 9, 7)
# We compare it to (3, 7) - if major >= 3 and minor >= 7, it's good!
if sys.version_info >= (3, 7):
    print("\n✓ Python version is 3.7 or higher - Good!")
    print("You can proceed to the next step.")
else:
    print("\n✗ Python version is too old. Please upgrade to 3.7+")
    print("Go to: https://www.python.org/downloads/")
```
