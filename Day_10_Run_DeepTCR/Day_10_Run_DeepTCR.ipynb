```python
# Import DeepTCR
# Note: DeepTCR_SS is for training, DeepTCR_WF is for inference/workflow
# Today we'll show training (DeepTCR_SS)

try:
    from DeepTCR.DeepTCR import DeepTCR_SS, DeepTCR_WF
except ImportError as exc:
    raise ImportError(
        "DeepTCR is not installed. Re-run Day 9 setup: `pip install DeepTCR`."
    ) from exc

from pathlib import Path

# Also import other libraries we'll need
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve, accuracy_score
from sklearn.preprocessing import LabelEncoder

sns.set(style="whitegrid")

print("✓ Imported libraries")
print("  - DeepTCR_SS / DeepTCR_WF: model training & inference")
print("  - sklearn: evaluation metrics & preprocessing")
print("  - matplotlib/seaborn: for plotting")
```

## Step 2: Initialize Model

**Biology context:**
- Create a new DeepTCR model
- This will learn from patient data

**CS context:**
- `DeepTCR_SS('model_name')` creates a new model
- Model name is used for saving/loading
- **What you know:** Like creating an object!

**Your Task:** Initialize DeepTCR model

```python
# Initialize DeepTCR model
# Using the workflow API so we can both train and run inference exactly as in the paper
model_name = "yost_day10_model"
model = DeepTCR_WF(model_name)

print("📦 DeepTCR model initialized")
print(f"  - Model directory: {model_name}")
print("  - Ready for data loading and training")
```

## Step 3: Load Data

**Biology context:**
- Load patient TCR files
- DeepTCR will process them
- **Note:** Data should be in 'data' subdirectory

**CS context:**
- `Get_Data()` loads TCR files
- `directory` = path to data folder
- `Load_Prev_Data=False` = start fresh
- **From source code:** Can specify column indices

**Your Task:** Load data in DeepTCR format

```python
# Load Yost cohort repertoires
if "model" not in globals() or model is None:
    raise RuntimeError("Model not initialized. Run Step 2 first.")

# Resolve paths relative to this notebook so it works from anywhere
yost_root = (Path("../data/DeepTCR_Cancer-master/Data/yost").resolve())
data_dir = yost_root / "data"
if not data_dir.exists():
    raise FileNotFoundError(
        f"Could not find Yost data directory at {data_dir}. Check your Day 9 setup."
    )

model.Get_Data(
    directory=str(data_dir),
    Load_Prev_Data=False,
    aa_column_beta=1,
    count_column=2,
    v_beta_column=7,
    d_beta_column=14,
    j_beta_column=21,
    data_cut=1.0
)

print("✓ Data loaded successfully")
print(f"  - Source: {data_dir}")
print(f"  - Samples discovered: {len(getattr(model, 'sample_list', []))}")
```

## Step 4: Prepare Response Labels

**Biology context:**
- Need to match patients to response labels
- Filter to pre-treatment samples (from source code!)
- **Why:** Only pre-treatment samples predict response

**CS context:**
- Load response.csv
- Match patient IDs
- Filter to pre-treatment
- **From source code:** Extract patient ID from filename

**Your Task:** Prepare response labels

```python
# Prepare response labels and align them with DeepTCR samples
response_file = Path("../data/DeepTCR_Cancer-master/Data/yost/response.csv").resolve()
if not response_file.exists():
    raise FileNotFoundError(f"Could not find response file at {response_file}")

response_df = pd.read_csv(response_file)
response_df["Patient"] = response_df["Patient"].str[:5]
response_df["Response"] = response_df["Response"].str.split().str[0]

sample_files = list(getattr(model, "sample_list", []))
if not sample_files:
    raise RuntimeError("No samples found on the model. Run Step 3 first.")

df_all = pd.DataFrame({"files": sample_files})
parts = df_all["files"].str.split("_", expand=True)
df_all["Patient"] = parts[0]
df_all["Tumor Type"] = parts[1]
df_all["time"] = parts[2].str.replace(r"\d+", "", regex=True)

df_merge = (
    df_all.merge(response_df, on=["Patient", "Tumor Type"], how="inner")
          .query("time == 'pre'")
          .copy()
)
df_merge["sample_id"] = (
    df_merge["Patient"] + "_" + df_merge["Tumor Type"] + "_" + df_merge["time"]
)

sample_dict = dict(zip(df_merge["files"], df_merge["sample_id"]))
label_dict = dict(zip(df_merge["files"], df_merge["Response"]))
label_dict2 = dict(zip(df_merge["sample_id"], df_merge["Response"]))
pre_sample_files = df_merge["files"].to_numpy()

print("✓ Response metadata prepared")
print(f"  - Total samples (pre-treatment): {len(df_merge)}")
print(f"  - Unique patients: {df_merge['Patient'].nunique()}")
```

## Step 5: Train Model

**Biology context:**
- Model learns which sequences predict response
- Uses attention to find important sequences
- **Goal:** Learn patterns from data

**CS context:**
- **From source code:** `Monte_Carlo_CrossVal()` trains the model (used in `Sup_Human_TCR.py`)
- **Alternative:** `Train_Supervised_Repertoire()` (simpler, less used in source code)
- **Key difference:** `Monte_Carlo_CrossVal()` does cross-validation automatically!
- **What you know:** Like training any ML model, but with built-in cross-validation!

**Your Task:** Train the DeepTCR model using `Monte_Carlo_CrossVal()`

```python
# Train DeepTCR with Monte Carlo cross-validation
if "model" not in globals() or model is None:
    raise RuntimeError("Model not initialized. Complete Steps 1-4 first.")

training_config = dict(
    folds=5,              # start small; set to 100 to reproduce the paper exactly
    LOO=6,
    epochs_min=10,
    size_of_net="small",
    num_concepts=64,
    combine_train_valid=True,
    hinge_loss_t=0.3,
    train_loss_min=0.1,
    seeds=np.arange(5),
    graph_seed=0,
)

print("🚀 Starting Monte Carlo Cross-Validation run")
print("  - This may take several minutes per fold (set folds=100 for final run)")
try:
    if hasattr(model, "use_hla"):
        model.use_hla = False
    model.Monte_Carlo_CrossVal(**training_config)
    print("✓ Training complete! Models saved inside the DeepTCR workspace")
except Exception as exc:
    raise RuntimeError(f"Training failed: {exc}")

```

## Step 6: Make Predictions

**Biology context:**
- Predict if patients will respond
- Get probabilities (0 to 1)
- **Goal:** See how well model works!

**CS context:**
- **From source code:** `Sample_Inference()` is used (from `yost.py`)
- Returns predictions in `Inference_Pred_Dict['crpr']`
- **Alternative:** `Predict_Repertoire()` (simpler, but less used)
- **Key:** Need to provide sequences and genes as arrays

**Your Task:** Make predictions using `Sample_Inference()`

```python
# Run Sample_Inference on the trained model outputs
required_vars = ["df_merge", "sample_dict", "label_dict", "label_dict2", "pre_sample_files"]
missing = [var for var in required_vars if var not in globals()]
if missing:
    raise RuntimeError(f"Missing metadata from Step 4: {missing}")

sample_ids = np.array(getattr(model, "sample_id", getattr(model, "sample_list", [])))
if sample_ids.size == 0:
    raise RuntimeError("Model does not expose sample IDs. Re-run Step 3.")

selected_mask = np.isin(sample_ids, pre_sample_files)
if not np.any(selected_mask):
    raise RuntimeError("No pre-treatment samples found in the loaded model.")

beta_sequences = model.beta_sequences[selected_mask]
v_beta = model.v_beta[selected_mask]
d_beta = model.d_beta[selected_mask]
j_beta = model.j_beta[selected_mask]

sample_labels = np.array([sample_dict[sid] for sid in sample_ids[selected_mask]])
class_labels = np.array([label_dict[sid] for sid in sample_ids[selected_mask]])

out = model.Sample_Inference(
    sample_labels=sample_labels,
    beta_sequences=beta_sequences,
    v_beta=v_beta,
    d_beta=d_beta,
    j_beta=j_beta,
)

df_preds = model.Inference_Pred_Dict["crpr"][:].copy()
df_preds["label"] = df_preds["Samples"].map(label_dict2)

print("✓ Predictions complete!")
print(f"  - Predictions DataFrame: {df_preds.shape}")
print(f"  - Preview columns: {list(df_preds.columns)}")
```

## Step 7: Evaluate Results - Calculate AUC

**Biology context:**
- AUC measures prediction quality
- AUC 0.82 = good prediction (paper result!)
- **Goal:** Reproduce paper's AUC!

**CS context:**
- `roc_auc_score(y_true, y_pred_proba)` calculates AUC
- Need probabilities, not binary predictions
- **What you know:** From sklearn!

**Your Task:** Calculate AUC

```python
# Evaluate AUC & accuracy
if "df_preds" not in globals():
    raise RuntimeError("Run Step 6 first to populate df_preds.")

df_eval = df_preds.dropna(subset=["label"]).copy()
if df_eval.empty:
    raise ValueError("Prediction frame has no labeled samples. Check Step 4 metadata.")

label_encoder = LabelEncoder()
df_eval["response_bin"] = label_encoder.fit_transform(df_eval["label"])
auc = roc_auc_score(df_eval["response_bin"], df_eval["Pred"])

threshold = 0.5
pred_binary = (df_eval["Pred"] >= threshold).astype(int)
accuracy = accuracy_score(df_eval["response_bin"], pred_binary)

print("📊 Evaluation Metrics")
print(f"  - AUC: {auc:.3f}")
print(f"  - Accuracy @ {threshold:.2f} threshold: {accuracy:.3f}")
print("  - Paper reference AUC ≈ 0.82 (100-fold CV)")
```

## Step 8: Evaluate Results - Bootstrapping (From Source Code!)

**Biology context:**
- Bootstrapping = resample data many times
- More robust evaluation
- **Why:** Shows confidence in results!

**CS context:**
- Resample data with replacement
- Calculate AUC each time
- **From source code:** yost.py uses 5000 bootstraps!
- **What you know:** Statistical resampling!

**Your Task:** Implement bootstrapping

```python
# Bootstrap AUC for confidence intervals
if "df_eval" not in globals():
    df_eval = df_preds.dropna(subset=["label"]).copy()
    df_eval["response_bin"] = LabelEncoder().fit_transform(df_eval["label"])

n_boots = 5000
bootstrap_scores = []
rng = np.random.default_rng(42)
for _ in range(n_boots):
    sample_idx = rng.integers(0, len(df_eval), len(df_eval))
    df_temp = df_eval.iloc[sample_idx]
    if df_temp["response_bin"].nunique() < 2:
        continue  # Skip degenerate samples
    bootstrap_scores.append(roc_auc_score(df_temp["response_bin"], df_temp["Pred"]))

bootstrap_scores = np.array(bootstrap_scores)
if bootstrap_scores.size == 0:
    raise RuntimeError("Bootstrap failed because resamples lacked both classes.")

print(f"📊 Bootstrap Results (n={n_boots}, valid={bootstrap_scores.size}):")
print(f"  - Mean AUC: {np.mean(bootstrap_scores):.3f}")
print(f"  - Std AUC: {np.std(bootstrap_scores):.3f}")
print(f"  - 95% CI: [{np.percentile(bootstrap_scores, 2.5):.3f}, {np.percentile(bootstrap_scores, 97.5):.3f}]")
```

## Step 9: Create ROC Curve (From Source Code!)

**Biology context:**
- ROC curve shows prediction performance
- X-axis = False Positive Rate
- Y-axis = True Positive Rate
- **Goal:** Visualize model performance!

**CS context:**
- `roc_curve()` creates ROC curve data
- `plt.plot()` draws the curve
- **From source code:** yost.py creates ROC plots!

**Your Task:** Create ROC curve

```python
# Plot ROC curve with bootstrap inset
if "df_eval" not in globals():
    raise RuntimeError("Need evaluation dataframe from Step 7.")

fig, ax = plt.subplots(figsize=(5, 5))
score = roc_auc_score(df_eval["response_bin"], df_eval["Pred"])
fpr, tpr, _ = roc_curve(df_eval["response_bin"], df_eval["Pred"])
ax.plot(fpr, tpr, lw=2, label=f"crpr (area = {score:0.4f})")
ax.plot([0, 1], [0, 1], color="navy", lw=2, linestyle="--", label="Random")
ax.set_xlabel("False Positive Rate", fontsize=12)
ax.set_ylabel("True Positive Rate", fontsize=12)
ax.set_xlim([0, 1])
ax.set_ylim([0, 1.05])
ax.legend(loc="lower right")
ax.set_title("ROC Curve - DeepTCR (Yost)")

if "bootstrap_scores" in globals() and bootstrap_scores.size:
    ax2 = fig.add_axes([0.58, 0.2, 0.3, 0.3])
    ax2.spines["right"].set_visible(False)
    ax2.spines["top"].set_visible(False)
    sns.violinplot(y=bootstrap_scores, ax=ax2, cut=0, color="skyblue")
    ax2.set_ylabel("AUC", fontsize=10)
    ax2.set_xticks([])
    ax2.hlines(y=0.5, xmin=-0.5, xmax=0.5, linewidth=2, linestyles="--", color="red")
    ax2.set_title("Bootstrap AUC", fontsize=10)

plt.tight_layout()
plt.show()
print("ROC curve rendered. Bootstrap inset reflects score distribution.")
```

## Step 10: Extract Attention Weights

**Biology context:**
- Attention weights show which sequences matter
- High weight = important sequence
- **Goal:** Understand which T cells predict response!

**CS context:**
- `Get_Attention_Weights()` returns attention weights
- Shows importance of each sequence
- **What you know:** Like feature importance!

**Your Task:** Extract and analyze attention weights

```python
# Extract attention weights and highlight top sequences
if "model" not in globals() or model is None:
    raise RuntimeError("Train the model first (Steps 1-5).")

attention_raw = model.Get_Attention_Weights()
if attention_raw is None:
    raise RuntimeError("Model did not return attention weights. Ensure training finished.")

weight_keys = [
    "weights",
    "weight",
    "attention",
    "attn",
    "attention_weights",
]

attention_df = None
if isinstance(attention_raw, dict):
    # Case 1: dict with explicit sequences + weights
    seq_values = attention_raw.get("beta_sequences") or attention_raw.get("sequences")
    key = next((k for k in weight_keys if k in attention_raw), None)
    if key is not None:
        attention_df = pd.DataFrame({
            "sequence": seq_values,
            "weight": np.asarray(attention_raw[key]).astype(float),
        })
    else:
        # Case 2: dict of sample_id -> weights
        rows = []
        for sample_id, weights in attention_raw.items():
            weights = np.asarray(weights).astype(float)
            rows.append(
                pd.DataFrame(
                    {
                        "sample_id": sample_id,
                        "weight": weights,
                        "sequence_index": np.arange(len(weights)),
                    }
                )
            )
        if rows:
            attention_df = pd.concat(rows, ignore_index=True)

if attention_df is None:
    # Fallback: treat raw output as array and pair with model sequences if available
    weights_np = np.asarray(attention_raw).astype(float)
    sequences_available = getattr(model, "beta_sequences", None)
    attention_df = pd.DataFrame({"weight": weights_np.ravel()})
    if sequences_available is not None and len(sequences_available) == len(attention_df):
        attention_df["sequence"] = sequences_available

attention_df = attention_df.dropna(subset=["weight"]).copy()
attention_df.sort_values("weight", ascending=False, inplace=True)

top_n = min(10, len(attention_df))
top_sequences = attention_df.head(top_n)
print(f"🔍 Retrieved {len(attention_df)} attention scores. Showing top {top_n}:")
print(top_sequences[[col for col in top_sequences.columns if col in {"sequence", "weight", "sample_id"}]].reset_index(drop=True))

# Keep handy for Day 11
attention_df_top = top_sequences.copy()
```
