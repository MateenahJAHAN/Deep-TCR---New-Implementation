## 🧬 Biology Context: Why Simple Baselines?

### The Problem We're Solving

**Simple explanation:**
- We have 50,000 sequences per patient
- We only know if the PATIENT responded (not individual sequences)
- **Question:** How do we predict response from 50,000 sequences?

### Why Try Simple Approaches First?

**Biology reason:**
- Simple approaches help us understand the problem
- They set a baseline (minimum performance)
- **Why it matters:** If simple approaches work, we don't need complex ones!
- **Paper finding:** Simple pooling doesn't work well - need attention!

### What Is Pooling?

**Simple explanation:**
- **Mean pooling:** Average all sequences → one vector per patient
- **Max pooling:** Take strongest sequence → one vector per patient
- **Why:** Convert 50,000 sequences → 1 vector (for standard ML)
- **Think:** Like summarizing a book into one sentence

---

## 💻 Computer Science Context: Machine Learning Basics

### What Is a Baseline?

**CS explanation:**
- **Baseline** = simple approach to compare against
- Like a "minimum bar" for performance
- **Why it matters:** If complex model doesn't beat baseline, it's not worth it!

### What Is Pooling?

**CS explanation:**
- **Pooling** = aggregating multiple items into one
- **Mean pooling:** `np.mean(array, axis=0)` - average along axis
- **Max pooling:** `np.max(array, axis=0)` - maximum along axis
- **Why:** Reduce dimensions (50k sequences → 1 vector)

### Why Use sklearn?

**CS reason:**
- sklearn has ready-made ML models
- Easy to use: `fit()` and `predict()`
- **What you know:** Like `RandomForestClassifier` from sklearn!

1. **Why do we need to pool sequences?**
   - We have 50,000 sequences per patient
   - Standard ML needs one vector per patient
   - Pooling converts many → one

2. **What's the difference between mean and max pooling?**

   - **Think:** Mean = democratic, Max = winner-takes-all

## Step 4: Try Max Pooling Baseline

**Biology context:**
- Max pooling = take strongest sequence
- **Think:** Like "winner takes all"
- **Problem:** Ignores all other sequences!

**CS context:**
- **Max pooling:** `np.max(array, axis=0)`
- Takes maximum along sequence axis
- **Result:** One vector per patient (strongest features)

**Your Task:** Implement max pooling and compare

```python
# Implement max pooling
# Max pooling: take strongest sequence for each patient

def max_pooling(sequences):
    """
    Max pooling: take maximum along sequence axis
    
    Input: (num_sequences, features) array
    Output: (features,) array
    """
    # ✅ Starter solution
    if sequences is not None and len(sequences) > 0:
        pooled = np.max(sequences, axis=0)
        return pooled
    else:
        return None
    
    # TODO: Replace starter code above with your own: pooled = np.max(sequences, axis=0)
```

```python
# Apply max pooling to all patients
# Loop through patient_sequences, apply max_pooling to each
# Store results in a list

# ✅ Starter solution
max_pooled_features = []

if patient_sequences is not None and len(patient_sequences) > 0:
    for sequences in patient_sequences:
        pooled = max_pooling(sequences)
        if pooled is not None:
            max_pooled_features.append(pooled)
        else:
            max_pooled_features.append(None)
    
    # Convert to numpy array
    if all(p is not None for p in max_pooled_features) and len(max_pooled_features) > 0:
        X_max = np.array(max_pooled_features)
        
        print(f"✓ Max pooled features:")
        print(f"  Shape: {X_max.shape}")
        print(f"  Expected: ({len(patient_sequences)}, {patient_sequences[0].shape[1]})")
        print(f"  Meaning: {X_max.shape[0]} patients, {X_max.shape[1]} features per patient")
        print(f"\nBiology: Each patient now represented by strongest sequence")
        print(f"CS: Ready for standard ML (one sample per patient)!")
    else:
        print("⚠️ Some pooling operations returned None. Complete the TODOs above!")
        X_max = None
else:
    print("⚠️ Complete the TODO in previous cell to create patient_sequences first!")
    X_max = None

# TODO: Replace starter code above with your own loop to apply max pooling
```

## Step 5: Understand Why Pooling Fails

**Biology context:**
- Not all sequences are equally important
- Some sequences predict response, others don't
- **Problem:** Pooling treats all sequences equally!
- **Solution:** Need attention to find important sequences

**CS context:**
- Pooling loses information about individual sequences
- Can't identify which sequences matter
- **Why it fails:** One-size-fits-all aggregation

**Your Task:** Think about the limitations

```python
# Visualize the problem
print("🔍 Why Pooling Fails:")
print(f"\nProblem:")
print(f"  - Patient has 50,000 sequences")
print(f"  - Only ~100 sequences might predict response")
print(f"  - Mean pooling: averages all 50,000 (dilutes signal!)")
print(f"  - Max pooling: takes only 1 (ignores others!)")

print(f"\nSolution (Day 8 - Attention):")
print(f"  - Learn which sequences are important")
print(f"  - Weight important sequences more")
print(f"  - Ignore irrelevant sequences")

print(f"\nPaper finding:")
print(f"  - Mean/Max pooling: AUC ~0.60 (barely better than random)")
print(f"  - DeepTCR with attention: AUC ~0.82 (much better!)")

print(f"\nBiology: Need to find the 'important' T cells!")
print(f"CS: Need attention mechanism to weight sequences!")
```
