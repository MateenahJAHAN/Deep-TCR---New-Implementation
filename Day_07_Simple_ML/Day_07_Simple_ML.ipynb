## 🧬 Biology Context: Why Simple Baselines?

### The Problem We're Solving

**Simple explanation:**
- We have 50,000 sequences per patient
- We only know if the PATIENT responded (not individual sequences)
- **Question:** How do we predict response from 50,000 sequences?

### Why Try Simple Approaches First?

**Biology reason:**
- Simple approaches help us understand the problem
- They set a baseline (minimum performance)
- **Why it matters:** If simple approaches work, we don't need complex ones!
- **Paper finding:** Simple pooling doesn't work well - need attention!

### What Is Pooling?

**Simple explanation:**
- **Mean pooling:** Average all sequences → one vector per patient
- **Max pooling:** Take strongest sequence → one vector per patient
- **Why:** Convert 50,000 sequences → 1 vector (for standard ML)
- **Think:** Like summarizing a book into one sentence

---

## 💻 Computer Science Context: Machine Learning Basics

### What Is a Baseline?

**CS explanation:**
- **Baseline** = simple approach to compare against
- Like a "minimum bar" for performance
- **Why it matters:** If complex model doesn't beat baseline, it's not worth it!

### What Is Pooling?

**CS explanation:**
- **Pooling** = aggregating multiple items into one
- **Mean pooling:** `np.mean(array, axis=0)` - average along axis
- **Max pooling:** `np.max(array, axis=0)` - maximum along axis
- **Why:** Reduce dimensions (50k sequences → 1 vector)

### Why Use sklearn?

**CS reason:**
- sklearn has ready-made ML models
- Easy to use: `fit()` and `predict()`
- **What you know:** Like `RandomForestClassifier` from sklearn!

## 🤔 Questions to Think About (Before Starting)

1. **Why do we need to pool sequences?**
   - We have 50,000 sequences per patient
   - Standard ML needs one vector per patient
   - Pooling converts many → one

2. **What's the difference between mean and max pooling?**
   - Mean: Average all sequences (all sequences matter equally)
   - Max: Take strongest sequence (only strongest matters)
   - **Think:** Mean = democratic, Max = winner-takes-all

3. **Why might pooling fail?**
   - Loses information about individual sequences
   - Can't find which sequences matter
   - **Biology:** Not all sequences are equally important!

- **Solution:** Need attention to find important sequences

**CS context:**
- Pooling loses information about individual sequences
- Can't identify which sequences matter
