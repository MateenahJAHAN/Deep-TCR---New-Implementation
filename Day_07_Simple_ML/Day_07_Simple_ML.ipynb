## 🧬 Biology Context: Why Simple Baselines?

### The Problem We're Solving

**Simple explanation:**
- We have 50,000 sequences per patient
- We only know if the PATIENT responded (not individual sequences)
- **Question:** How do we predict response from 50,000 sequences?

### Why Try Simple Approaches First?

**Biology reason:**
- Simple approaches help us understand the problem
- They set a baseline (minimum performance)
- **Why it matters:** If simple approaches work, we don't need complex ones!
- **Paper finding:** Simple pooling doesn't work well - need attention!

### What Is Pooling?

**Simple explanation:**
- **Mean pooling:** Average all sequences → one vector per patient
- **Max pooling:** Take strongest sequence → one vector per patient
- **Why:** Convert 50,000 sequences → 1 vector (for standard ML)
- **Think:** Like summarizing a book into one sentence

## 💻 Computer Science Context: Machine Learning Basics

### What Is a Baseline?

**CS explanation:**
- **Baseline** = simple approach to compare against
- Like a "minimum bar" for performance
- **Why it matters:** If complex model doesn't beat baseline, it's not worth it!

### What Is Pooling?

**CS explanation:**
- **Pooling** = aggregating multiple items into one
- **Mean pooling:** `np.mean(array, axis=0)` - average along axis
- **Max pooling:** `np.max(array, axis=0)` - maximum along axis
- **Why:** Reduce dimensions (50k sequences → 1 vector)

### Why Use sklearn?

**CS reason:**
- sklearn has ready-made ML models
- Easy to use: `fit()` and `predict()`
- **What you know:** Like `RandomForestClassifier` from sklearn!

````python
# Implement mean pooling
# Mean pooling: average all sequences for each patient

def mean_pooling(sequences):
    """
    Mean pooling: average all sequences
    
    Input: (num_sequences, features) or (num_sequences, L, C)
    Output: 1D array suitable for sklearn (features,)
    """
    if sequences is not None and len(sequences) > 0:
        pooled = np.mean(sequences, axis=0)
        return pooled.flatten()
    return None

# Test with first patient
if patient_sequences is not None and len(patient_sequences) > 0:
    test_sequences = patient_sequences[0]
    print(f"Testing mean pooling:")
    print(f"  Input shape: {test_sequences.shape}")
    
    pooled = mean_pooling(test_sequences)
    
    if pooled is not None:
        print(f"  Output shape: {pooled.shape}")
        print(f"  ✓ Mean pooling works!")
        print(f"\nBiology: Averaged all sequences → lost individual information")
        print(f"CS: Reduced dimensions: {test_sequences.shape} → {pooled.shape}")
    else:
        print("  ⚠️ Complete the TODO in mean_pooling() function above!")
        print("  Hint: Use np.mean(sequences, axis=0)")
else:
    print("⚠️ Complete the TODO in previous cell to create patient_sequences first!")
````

````python
# Apply mean pooling to all patients
# Loop through patient_sequences, apply mean_pooling to each
# Store results in a list

# ✅ Starter solution
mean_pooled_features = []

if patient_sequences is not None and len(patient_sequences) > 0:
    for sequences in patient_sequences:
        pooled = mean_pooling(sequences)
        if pooled is not None:
            mean_pooled_features.append(pooled)
        else:
            mean_pooled_features.append(None)
    
    # Convert to numpy array
    if all(p is not None for p in mean_pooled_features) and len(mean_pooled_features) > 0:
        X_mean = np.array(mean_pooled_features)
        
        print(f"✓ Mean pooled features:")
        print(f"  Shape: {X_mean.shape}")
        print(f"  Expected: ({len(patient_sequences)}, {patient_sequences[0].shape[1]})")
        print(f"  Meaning: {X_mean.shape[0]} patients, {X_mean.shape[1]} features per patient")
        print(f"\nBiology: Each patient now represented by one averaged vector")
        print(f"CS: Ready for standard ML (one sample per patient)!")
    else:
        print("⚠️ Some pooling operations returned None. Complete the TODOs above!")
        X_mean = None
else:
    print("⚠️ Complete the TODO in previous cell to create patient_sequences first!")
    X_mean = None

# TODO: Replace starter code above with your own loop to apply mean pooling
````

### Mateenah’s result decoder
- **Target range:** accuracy ≈ 0.5–0.7, AUC ≈ 0.55–0.62 (this matches Figure 2 commentary in the paper).

````python
# Evaluate mean-pooling baseline across multiple random splits for stability
if X_mean is not None and patient_labels is not None:
    seeds = [0, 1, 2, 3, 4]
    accs = []
    aucs = []
    for seed in seeds:
        X_tr, X_te, y_tr, y_te = train_test_split(
            X_mean, patient_labels, test_size=0.3, random_state=seed, stratify=patient_labels
        )
        clf = RandomForestClassifier(n_estimators=100, random_state=seed)
        clf.fit(X_tr, y_tr)
        preds = clf.predict(X_te)
        probs = clf.predict_proba(X_te)[:, 1]
        accs.append(accuracy_score(y_te, preds))
        if len(set(y_te)) >= 2:
            aucs.append(roc_auc_score(y_te, probs))
    accuracy_mean = float(np.mean(accs)) if accs else float("nan")
    auc_mean = float(np.mean(aucs)) if aucs else float("nan")
    print("📊 Mean Pooling Results (average over 5 splits)")
    print(f"  Accuracy: {accuracy_mean:.3f}")
    print(f"  AUC: {auc_mean if auc_mean==auc_mean else 'nan'}")
    if not aucs:
        print("  ⚠️ Some splits lacked both classes in y_test; try more patients for smoother AUC.")
    
    print("\nBiology interpretation:")
    print("  - The paper reports AUC ≈0.60 once all 15 patients are included; with ~10 patients expect noisy swings (0.2–0.6).")
    print("  - Anything well above 0.7 usually means label leakage or overly tiny splits.")
    
    print("\nCS interpretation:")
    print("  - Mean pooling treats all sequences equally, so predictive motifs get diluted.")
    print("  - This poor baseline justifies using attention (Day 8).")
else:
    print("⚠️ Need X_mean and patient_labels before evaluating mean pooling. Run the earlier cells.")
    accuracy_mean = None
    auc_mean = None
````

## Step 4: Try Max Pooling Baseline

**Biology context:**
- Max pooling = take strongest sequence
- **Think:** Like "winner takes all"
- **Problem:** Ignores all other sequences!

**CS context:**
- **Max pooling:** `np.max(array, axis=0)`
- Takes maximum along sequence axis
- **Result:** One vector per patient (strongest features)

**Your Task:** Implement max pooling and compare

````python
# Implement max pooling
# Max pooling: take strongest sequence for each patient

def max_pooling(sequences):
    """
    Max pooling: take maximum along sequence axis
    
    Input: (num_sequences, features) or (num_sequences, L, C)
    Output: 1D array suitable for sklearn
    """
    if sequences is not None and len(sequences) > 0:
        pooled = np.max(sequences, axis=0)
        return pooled.flatten()
    return None
````

````python
# Apply max pooling to all patients
# Loop through patient_sequences, apply max_pooling to each
# Store results in a list

# ✅ Starter solution
max_pooled_features = []

if patient_sequences is not None and len(patient_sequences) > 0:
    for sequences in patient_sequences:
        pooled = max_pooling(sequences)
        if pooled is not None:
            max_pooled_features.append(pooled)
        else:
            max_pooled_features.append(None)
    
    # Convert to numpy array
    if all(p is not None for p in max_pooled_features) and len(max_pooled_features) > 0:
        X_max = np.array(max_pooled_features)
        
        print(f"✓ Max pooled features:")
        print(f"  Shape: {X_max.shape}")
        print(f"  Expected: ({len(patient_sequences)}, {patient_sequences[0].shape[1]})")
        print(f"  Meaning: {X_max.shape[0]} patients, {X_max.shape[1]} features per patient")
        print(f"\nBiology: Each patient now represented by strongest sequence")
        print(f"CS: Ready for standard ML (one sample per patient)!")
    else:
        print("⚠️ Some pooling operations returned None. Complete the TODOs above!")
        X_max = None
else:
    print("⚠️ Complete the TODO in previous cell to create patient_sequences first!")
    X_max = None

# TODO: Replace starter code above with your own loop to apply max pooling
````

````python
# Train + evaluate max-pooling baseline (average over multiple splits)
if X_max is not None and patient_labels is not None:
    seeds = [10, 11, 12, 13, 14]
    accs = []
    aucs = []
    for seed in seeds:
        X_tr, X_te, y_tr, y_te = train_test_split(
            X_max, patient_labels, test_size=0.3, random_state=seed, stratify=patient_labels
        )
        clf = RandomForestClassifier(n_estimators=100, random_state=seed)
        clf.fit(X_tr, y_tr)
        preds = clf.predict(X_te)
        probs = clf.predict_proba(X_te)[:, 1]
        accs.append(accuracy_score(y_te, preds))
        if len(set(y_te)) >= 2:
            aucs.append(roc_auc_score(y_te, probs))
    accuracy_max = float(np.mean(accs)) if accs else float("nan")
    auc_max = float(np.mean(aucs)) if aucs else float("nan")
    print("📊 Max Pooling Results (average over 5 splits)")
    print(f"  Accuracy: {accuracy_max:.3f}")
    print(f"  AUC: {auc_max if auc_max==auc_max else 'nan'}")
    if not aucs:
        print("  ⚠️ Some splits lacked both classes; try saving more patients in Day 6.")
else:
    print("⚠️ Create X_max and patient_labels before training the max-pooling baseline.")
    accuracy_max = None
    auc_max = None
````
