## 📚 How to Use Jupyter Notebooks (Basic Guide)

### Understanding Notebook Cells

A notebook has two types of cells:

1. **Markdown Cells** (like this one)
   - Contains text, explanations, instructions
   - Not executable code
   - Used for documentation

2. **Code Cells** (like the one below)
   - Contains Python code
   - Can be executed
   - Shows output below the cell

### Basic Operations

**To run a cell:**
- Click the cell to select it
- Press `Shift + Enter` (runs cell and moves to next)
- Or click the "Run" button (▶) in the toolbar

**To add a new cell:**
- Press `A` (adds cell above)
- Press `B` (adds cell below)
- Or click "+ Code" or "+ Markdown" button

**To edit a cell:**
- Double-click on a markdown cell to edit
- Click on a code cell to edit

**To save:**
- Press `Ctrl + S` (Windows/Linux) or `Cmd + S` (Mac)
- Or click "File" → "Save"
