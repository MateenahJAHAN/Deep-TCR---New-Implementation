## 📚 How to Use Jupyter Notebooks (Quick Guide)

### Understanding Cells

**Two types of cells:**

1. **Markdown Cells** (like this one)
   - Contains text, explanations, instructions
   - Not executable code
   - **How to recognize:** No `[ ]` or `[1]` on the left

2. **Code Cells** (like the one below)
   - Contains Python code
   - Can be executed
   - Shows output below when run
   - **How to recognize:** Has `[ ]` or `[1]` on the left

### Basic Operations

**To run a cell:**
- **VS Code:** Click the "Run Cell" button (▶) above the cell, or press `Shift+Enter`
- **Colab:** Click the play button (▶) on the left, or press `Shift+Enter`
- **Watch:** The `[ ]` changes to `[1]` when you run it (number shows execution order)

**To add a new cell:**
- **VS Code:** Click "+ Code" or "+ Markdown" button in toolbar
- **Colab:** Click "+ Code" or "+ Text" button

**To edit a cell:**
- **Markdown:** Double-click on the cell to edit
- **Code:** Just click on it and start typing

**To save:**
- **VS Code:** Press `Ctrl+S` (Windows/Linux) or `Cmd+S` (Mac)
- **Colab:** Auto-saves to your Google Drive!

## 🐍 Basic Python Review (If You Need It)

**Don't worry if you're new to Python!** Here's what you need to know:

### Variables (Storing Information)
```python
name = "Alice"      # String (text) - notice the quotes
age = 25            # Integer (whole number)
height = 5.6        # Float (decimal number)
```

### Print Statement (Showing Output)
```python
print("Hello, World!")                    # Print text
print(name)                                # Print variable
print(f"My name is {name} and I'm {age}")  # f-string (combines text + variables)
```

### Lists (Collections of Items)
```python
numbers = [1, 2, 3, 4, 5]      # List of numbers
print(numbers[0])              # First item: 1 (starts at 0!)
print(len(numbers))            # Length: 5
```
