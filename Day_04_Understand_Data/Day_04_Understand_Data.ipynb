## 🧬 Biology Context: Understanding Multiple Patients

### What Is a Repertoire?

**Simple explanation:**
- A patient's TCR repertoire = all their T cell barcodes
- Like a library of all T cells in that patient
- **Repertoire size** = how many unique TCR sequences
- **Why it matters:** Larger repertoires might mean better immune response!

### Why Compare Multiple Patients?

**Biology reason:**
- Each patient has a unique immune system
- Some patients respond to treatment, some don't
- **Question:** Do responders have different TCR patterns?
- **Paper finding:** Yes! Responders have different V gene usage patterns

### What Are Responders vs Non-Responders?

**Simple explanation:**
- **Responders:** Patients whose tumors shrank after immunotherapy ✅
- **Non-responders:** Patients whose tumors didn't shrink ❌
- **Why it matters:** We want to predict who will respond BEFORE treatment!

## Step 1: Load Multiple Patient Files

**From Roadmap:** Load 5 different patient files

**Biology context:**
- Each file = one patient's TCR repertoire
- We'll analyze multiple patients to find patterns
- **Goal:** Understand how repertoires differ between patients

**CS context:**
- We'll use a `for` loop to process each file
- Store results in a list or dictionary
- **What you know:** Just like iterating through a list!

**Your Task:** Load multiple patient files and count sequences for each

```python
# Import libraries
import pandas as pd
from pathlib import Path

# Locate data directory (works for repo + Colab once data uploaded)
data_dir = Path("../data/DeepTCR_Cancer-master/Data/yost/data")
if not data_dir.exists():
    for alt in [Path("data/DeepTCR_Cancer-master/Data/yost/data"), Path("./data/DeepTCR_Cancer-master/Data/yost/data")]:
        if alt.exists():
            data_dir = alt
            print(f"✓ Found data at: {alt}")
            break

# ✅ Starter listing
if ("all_files" not in globals()) or (all_files is None):
    all_files = sorted(data_dir.glob("*.tsv"))
    print("✓ Automatically listed patient files.")
else:
    print("✓ Using your custom `all_files` list.")

print(f"Found {len(all_files)} patient files")
print("\nFirst 5 files:")
for f in all_files[:5]:
    print(f"  - {f.name}")

if ("selected_files" not in globals()) or (selected_files is None):
    selected_files = all_files[:5]
    print("\n✓ Selected the first 5 files as a starting point.")
else:
    print("\n✓ Using your custom `selected_files` selection.")

print(f"\n✓ Ready to analyze {len(selected_files)} patients")
```

```python
# Count productive sequences for each selected patient file
if ("selected_files" not in globals()) or not selected_files:
    raise ValueError("Make sure `selected_files` is defined in the previous cell.")

if ("patient_sizes" not in globals()) or not patient_sizes:
    patient_sizes = {}

for file_path in selected_files:
    patient_name = file_path.stem
    if patient_name in patient_sizes and patient_sizes[patient_name] is not None:
        continue  # keep learner's result

    if not file_path.exists():
        print(f"⚠️ Skipping missing file: {file_path}")
        patient_sizes[patient_name] = None
        continue

    df = pd.read_csv(file_path, sep="\t")
    productive = df[df["sequenceStatus"] == "In"]
    num_sequences = len(productive)
    patient_sizes[patient_name] = num_sequences
    print(f"  {patient_name}: {num_sequences:,} sequences (productive)")

print(f"\n✓ Processed {len([v for v in patient_sizes.values() if v is not None])} patients")
print("Patient repertoire sizes:")
for patient, size in patient_sizes.items():
    if size is not None:
        print(f"  {patient}: {size:,}")
```

## Step 2: Calculate Average Repertoire Size

**From Roadmap:** Calculate average repertoire size

**Biology context:**
- Average repertoire size = how many unique TCRs patients typically have
- **Why it matters:** Larger repertoires might mean better immune diversity
- **Paper finding:** Responders often have more diverse repertoires

**CS context:**
- Calculate mean, median, min, max from our dictionary
- Use `np.mean()`, `np.median()`, etc.
- **What you know:** Just like calculating statistics from a list!

**Your Task:** Calculate statistics about repertoire sizes

```python
import numpy as np

sizes_list = [size for size in patient_sizes.values() if size is not None]
if not sizes_list:
    raise ValueError("`patient_sizes` is empty. Run the previous cell to populate it.")

avg_size = float(np.mean(sizes_list))
median_size = float(np.median(sizes_list))
min_size = int(np.min(sizes_list))
max_size = int(np.max(sizes_list))

print(f"Repertoire sizes: {sizes_list}")
print("\n📊 Repertoire Size Statistics:")
print(f"  Average (mean): {avg_size:,.0f} sequences")
print(f"  Median: {median_size:,.0f} sequences")
print(f"  Minimum: {min_size:,} sequences")
print(f"  Maximum: {max_size:,} sequences")

print("\nBiology interpretation:")
print(f"  - Patients have {avg_size:,.0f} unique TCR sequences on average")
print("  - This represents their immune repertoire diversity")
print("  - Higher diversity = more T cell types = potentially better response")
```

## Step 3: Find Most Common V Genes Across All Patients

**From Roadmap:** Find most common V genes across all patients

**Biology context:**
- V genes determine part of the TCR structure
- Some V genes are used more often across all patients
- **Why it matters:** Common V genes might be important for immune function
- **Paper finding:** Certain V genes are associated with responders

**CS context:**
- Collect V genes from all patients
- Combine into one list
- Count frequencies using `value_counts()`
- **What you know:** Just like counting values in one DataFrame, but across multiple!

**Your Task:** Find which V genes are most common across all patients

```python
# Collect V genes from all selected patients
if ("selected_files" not in globals()) or not selected_files:
    raise ValueError("Run the earlier cells to pick patient files first.")

if ("all_v_genes" not in globals()) or not all_v_genes:
    all_v_genes = []

for file_path in selected_files:
    df = pd.read_csv(file_path, sep="\t")
    productive = df[df["sequenceStatus"] == "In"]
    all_v_genes.extend(productive["vGeneName"].dropna().tolist())

print(f"\n✓ Collected {len(all_v_genes):,} V gene entries from {len(selected_files)} patients")
print("  (Some V genes appear multiple times per patient)")
```

```python
import pandas as pd

if not all_v_genes:
    raise ValueError("Run the previous cell to collect V genes first.")

v_gene_counts = pd.Series(all_v_genes).value_counts()
top_v_genes = v_gene_counts.head(10)

print("\n🔬 Top 10 Most Common V Genes Across All Patients:")
print(f"\n{'V Gene':<20} {'Count':<15} {'Percentage':<15}")
print("-" * 50)
for v_gene, count in top_v_genes.items():
    percentage = (count / len(all_v_genes)) * 100
    print(f"{v_gene:<20} {count:<15,} {percentage:>6.2f}%")

print("\nBiology interpretation:")
print("  - These V genes are used most frequently across patients")
print("  - Common V genes might be important for immune function")
print("  - Paper found: Certain V gene patterns predict response!")
```

## Step 4: Plot Sequence Length Distributions

**From Roadmap:** Plot sequence length distributions

**Biology context:**
- TCR sequences have different lengths
- Most sequences are 10-25 amino acids
- **Why it matters:** Length distribution tells us about repertoire characteristics
- **Paper finding:** Length distributions are similar between responders/non-responders

**CS context:**
- Use matplotlib to create histograms
- Histogram = bar chart showing how many sequences of each length
- **What you know:** Like Excel histograms, but in Python!

**Your Task:** Create a histogram showing sequence length distribution

```python
import matplotlib.pyplot as plt

if not selected_files:
    raise ValueError("Select patient files earlier in the notebook.")

first_file = selected_files[0]
df_lengths = pd.read_csv(first_file, sep="\t")
productive_lengths = df_lengths[df_lengths["sequenceStatus"] == "In"]
lengths = productive_lengths["aminoAcid"].str.len().tolist()

print(f"✓ Collected {len(lengths)} sequence lengths from {first_file.name}")
print(f"  Length range: {min(lengths)} to {max(lengths)} amino acids")
print(f"  Average length: {np.mean(lengths):.1f} amino acids")
```

```python
plt.figure(figsize=(6, 4))
plt.hist(lengths, bins=20, color="#4f7cac", edgecolor="white")
plt.xlabel("Sequence Length (amino acids)")
plt.ylabel("Number of Sequences")
plt.title("TCR Sequence Length Distribution")
plt.tight_layout()
plt.show()

print("\n✓ Histogram created!")
print("\nBiology interpretation:")
print("  - Most sequences are 10-25 amino acids (typical range)")
print("  - Very short or very long sequences are rare")
print("  - This distribution is typical for TCR repertoires")
```

## Step 6: Compare Responders vs Non-Responders

**Biology context:**
- This is the KEY question: Do responders have different TCR patterns?
- **Paper finding:** Yes! Responders have different V gene usage and repertoire diversity
- **Why it matters:** If we can find patterns, we can predict response!

**CS context:**
- We'll compare statistics between two groups
- Use filtering to separate groups
- Create visualizations to compare
- **What you know:** Like comparing two groups in Excel!

**Your Task:** Compare repertoire sizes between responders and non-responders

```python
# Build comparison DataFrame with repertoire size + response label
if "response_df" not in globals():
    raise ValueError("Load response data before creating the comparison table.")

comparison_data = []
for file_path in selected_files:
    patient_name = file_path.stem
    patient_id = patient_name.split("_")[0]

    df = pd.read_csv(file_path, sep="\t")
    productive = df[df["sequenceStatus"] == "In"]
    num_sequences = len(productive)

    response_row = response_df[response_df["Patient"] == patient_id]
    response_value = response_row["Response"].iloc[0] if not response_row.empty else "Unknown"

    comparison_data.append({
        "Patient": patient_id,
        "File": file_path.name,
        "Repertoire_Size": num_sequences,
        "Response": response_value,
    })

comparison_df = pd.DataFrame(comparison_data)
print("\n✓ Created comparison DataFrame:")
print(comparison_df.head())
```

```python
responders = comparison_df[comparison_df["Response"] == "Yes"]
non_responders = comparison_df[comparison_df["Response"] == "No"]

avg_responders = responders["Repertoire_Size"].mean()
avg_non_responders = non_responders["Repertoire_Size"].mean()

difference = avg_responders - avg_non_responders

print("\n📊 Comparison: Responders vs Non-Responders")
print("\n  Responders:")
print(f"    Average repertoire size: {avg_responders:,.0f} sequences")
print(f"    Number of patients: {len(responders)}")
print("\n  Non-Responders:")
print(f"    Average repertoire size: {avg_non_responders:,.0f} sequences")
print(f"    Number of patients: {len(non_responders)}")
print(f"\n  Difference: {difference:,.0f} sequences")

print("\nBiology interpretation:")
if difference > 0:
    print("  - Responders have larger repertoires on average")
    print("  - This suggests diversity might be important for response!")
else:
    print("  - Non-responders have larger repertoires")
    print("  - Or: No clear difference (need more data)")
```

```python
import seaborn as sns

sns.set_style("whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=comparison_df, x="Response", y="Repertoire_Size", ax=axes[0], palette="pastel")
sns.barplot(data=comparison_df, x="Response", y="Repertoire_Size", ax=axes[1], palette="pastel", ci=95)

axes[0].set_title('Repertoire Size Distribution')
axes[0].set_ylabel('Repertoire Size (sequences)')
axes[1].set_title('Average Repertoire Size')
axes[1].set_ylabel('Average Repertoire Size (sequences)')

plt.tight_layout()
plt.show()

print("\n✓ Comparison plots created!")
print("\nBiology interpretation:")
print("  - Box plot shows the distribution of repertoire sizes")
print("  - Bar chart shows the average for each group")
print("  - If boxes don't overlap much, there's a real difference!")
```
