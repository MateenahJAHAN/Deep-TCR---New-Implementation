## 🧬 Biology Context: Why DeepTCR?

### What Is DeepTCR?

**Simple explanation:**
- DeepTCR = Deep learning for T-cell Receptor analysis
- It's a Python package that implements the paper's method
- **Why it matters:** Saves us from implementing everything from scratch!

### What Does DeepTCR Do?

**Biology context:**
- Takes patient TCR sequences
- Learns which sequences predict response
- Uses attention to find important sequences
- **Result:** Predicts if patient will respond to immunotherapy

### Why Use the Package?

**Biology reason:**
- Implements the paper's method correctly
- Handles all the complex parts (encoding, attention, etc.)
- **Why it matters:** We can focus on using it, not building it!

**CS context:**

- **What you know:** Like calling functions!

**Your Task:** Learn about DeepTCR methods

```python
print("📚 DeepTCR API Overview:")
print("\n" + "="*60)

print("\n1️⃣ Initialize Model:")
print("   model = DeepTCR_SS('model_name')")
print("   - Creates a new DeepTCR model")
print("   - 'model_name' = name for saving/loading")

print("\n2️⃣ Load Data:")
print("   model.Get_Data(directory='path/to/data')")
print("   - Loads patient TCR files")
print("   - Loads response labels")
print("   - Prepares data for training")

print("\n3️⃣ Train Model:")
print("   model.Train_Supervised_Repertoire(epochs=100)")
print("   - Trains the model")
print("   - Uses attention mechanism")
print("   - Learns which sequences matter")

print("\n4️⃣ Make Predictions:")
print("   predictions = model.Predict_Repertoire()")
print("   - Predicts patient response")
print("   - Returns probabilities")

print("\n5️⃣ Get Attention Weights:")
print("   attention = model.Get_Attention_Weights()")
print("   - Shows which sequences are important")
print("   - Helps understand the model")

print("\n" + "="*60)
print("\nBiology: Simple API for complex biology problem!")
print("CS: Just a few method calls - easy to use!")
```
